```python
import numpy as np
import pandas as pd

# Define data with missing (NaN) and duplicate values
data = {
    "A": [10, 20, np.nan, 10, 50],
    "B": [5.5, np.nan, 7.2, 5.5, 9.1],
    "C": ["Apple", "Banana", "Cherry", "Apple", "Elderberry"],
    "D": [True, False, np.nan, True, False],
}

# Create a 5-row, 4-column DataFrame

print(data)
```

```python
class DataValidator:

    def __init__(self, data):

        self.data = pd.DataFrame(data) if isinstance(data, dict) else data

        self.report = {
            "result": {}
        }


    def missing_values(self):

        missing = self.data.isnull().sum()

        return missing[missing > 0]


    def duplicate_rows(self):

        duplicates = self.data.duplicated().sum()

        return duplicates


    def data_types(self):

        data_type = self.data.dtypes

        return data_type


    def shape(self):

        row, columns = self.data.shape

        return {
            "rows": row,
            "column": columns
        }


    def constant_column(self):

        unique = self.data.nunique()

        constant_column = unique[unique == 1].index.tolist()

        return constant_column


    def infinite_values(self):

        column_selection = self.data.select_dtypes(include=[np.number])

        infinite_values = np.isinf(column_selection).sum()

        return infinite_values


    def numeric_summery(self):

        column_selection = self.data.select_dtypes(include=[np.number])

        numeric_summery = column_selection.describe()

        return numeric_summery


    def add_result(self, name, value, severity):

        self.report["result"][name] = {
            "value": value,
            "severity": severity
        }


    def get_status(self):

        for result in self.report["result"].values():

            if result["severity"] == "FAIL":
                return "FAIL"


        for result in self.report["result"].values():

            if result["severity"] == "WARNING":
                return "WARNING"


        return "PASS"


    def validate(self):

        # Empty dataset
        if self.data.empty:

            self.add_result(
                "empty_dataset",
                True,
                "FAIL"
            )

        else:

            self.add_result(
                "empty_dataset",
                False,
                "INFO"
            )


        # Duplicate rows
        duplicate_rows = self.duplicate_rows()

        self.add_result(
            "duplicate_rows",
            duplicate_rows,
            "WARNING" if duplicate_rows > 0 else "INFO"
        )


        # Missing values
        missing_values = self.missing_values()

        self.add_result(
            "missing_values",
            missing_values,
            "WARNING" if len(missing_values) > 0 else "INFO"
        )


        # Constant columns
        constant_column = self.constant_column()

        self.add_result(
            "constant_column",
            constant_column,
            "WARNING" if len(constant_column) > 0 else "INFO"
        )


        # Infinite values
        infinite_values = self.infinite_values()

        self.add_result(
            "infinite_values",
            infinite_values,
            "WARNING" if infinite_values.sum() > 0 else "INFO"
        )


        # Information
        self.add_result(
            "data_types",
            self.data_types(),
            "INFO"
        )

        self.add_result(
            "shape",
            self.shape(),
            "INFO"
        )

        self.add_result(
            "numeric_summary",
            self.numeric_summery(),
            "INFO"
        )


        # Final status
        self.report["status"] = self.get_status()

        return self.report
    
validator = DataValidator(data)

report = validator.validate()

print(report)
```
