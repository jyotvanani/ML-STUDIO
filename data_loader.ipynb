```python
# BaseaLoader
#     ↓
# Defines the RULE
# "Every loader must have load()"

# CSVLoader / ExcelLoader
#     ↓
# Does the ACTUAL WORK
# "Here's how I load this particular format"

# DataLoader
#     ↓
# Provides the COMMON ENTRY POINT
# "Give me the data"
```

```python
import pandas as pd
from abc import ABC , abstractmethod 
```

```python
class base_loader(ABC):#baseloader does not do anything but it is a base class for other loaders to inherit from
    @abstractmethod
    def load(self):
        pass          
```

```python
class csv_loader(base_loader):
    def __init__(self , file_path):
        self.file_path = file_path
    def load(self):
        return pd.read_csv(self.file_path)    
```

```python
class excel_loader(base_loader):
    def __init__(self , file_path):
        self.file_path = file_path
    def load(self):
        return pd.read_excel(self.file_path)    
```

```python
class Json_loader(base_loader):
    def __init__(self, path):
        self.path = path
    def load(self):
        return pd.read_json(self.path)
```

```python
Loader_registry = {
    ".csv" : csv_loader,
    ".excel" : excel_loader,
    ".Json" : Json_loader

}
```

```python
from pathlib import Path

class dataloader:
    def __init__(self ,path , registry):    
        self.path = Path(path)
        self.registry = registry

    def select_loader(self):
        extension = self.path.suffix.lower()

        loader_class = self.Loader_registry.get(extension)

        if loader_class is None:
            raise ValueError(f"unsupported file type: {extension}")

        return loader_class(self.path)

                
    def load(self):

          if not self.path.exists():
                    raise FileNotFoundError(f"file not found : {self.path}")
          
          loader = self.select_loader()
          return loader.load()   
```
