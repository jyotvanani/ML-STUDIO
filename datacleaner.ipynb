```python
import numpy as np
import pandas as pd
```

```python
class DataCleaner:
    def __init__(self , data):
        self.data = data
        
    def Missing_values(self):
        
        numeric_columns = self.data.select_dtypes(include=[np.number]).columns
        for column in numeric_columns:
            if self.data[column].isnull().any():
                mean_value = self.data[column].mean()
                self.data[column].fillna(mean_value , inplace=True)  #work of inplace = true is to modify the original dataframe instead of creating a new one
                
        categorical_columns = self.data.select_dtypes(include=['object']).columns #why put the columns() at the last? because we want to get the column names of the dataframe
        for column in categorical_columns:
            if self.data[column].isnull().any():
                mode_value = self.data[column].mode()[0] #why put the [0] at the last? because mode() returns a series, we want to get the first value of the series
                self.data[column].fillna(mode_value , inplace=True) 
           
        return self.data         
                
    def Remove_duplicates(self):
        self.data = self.data.drop_duplicates()
        return self.data
    
    def Fix_data_types(self , column , new_type):
        if column in self.data.columns:
            try:
               self.data[column] = self.data[column].astype(new_type)
            except ValueError:
                print(f'cannot convert {column} to {new_type}')
                
                
    def clean_data(self):
        self.Missing_values()
        self.Remove_duplicates()
        return self.data
                    
                            
```
